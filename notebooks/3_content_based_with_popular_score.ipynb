```python
def PtopK(y_true, y_pred):
    """
    Precision at K: Tính tỷ lệ phần trăm của các mục đúng trong top K dự đoán.
    """
    if y_true in y_pred:
        return 1 / len(y_pred)
    else:
        return 0


def RtopK(y_true, y_pred):
    """
    Recall at K: Tính tỷ lệ phần trăm của các mục đúng được tìm thấy trong top K dự đoán.
    """
    if y_true in y_pred:
        return 1
    else:
        return 0


def MRR(y_true, y_pred):
    """
    Mean Reciprocal Rank (MRR): Tính trung bình của các hạng mục đúng trong top K dự đoán.
    """
    if y_true not in y_pred:
        return 0

    rank = 0
    for index, value in enumerate(y_pred):
        if value == y_true:
            rank = 1 / (index + 1)
            break

    return rank


def NDCG(y_true, y_pred, k=None):
    """
    Normalized Discounted Cumulative Gain (NDCG): Đo lường chất lượng của các dự đoán xếp hạng.
    """
    if y_true not in y_pred:
        return 0.0

    true_index = y_pred.index(y_true)
    ideal_sorted = sorted(y_pred, reverse=True)
    ideal_index = ideal_sorted.index(y_true)

    dcg = 1.0 / np.log2(true_index + 2)
    idcg = 1.0 / np.log2(ideal_index + 2)

    ndcg = dcg / idcg

    if k is not None:
        ideal_sorted = ideal_sorted[:k]
        y_pred = y_pred[:k]
        true_index = y_pred.index(y_true)
        ideal_index = ideal_sorted.index(y_true)
        dcg = 1.0 / np.log2(true_index + 2)
        idcg = 1.0 / np.log2(ideal_index + 2)
        ndcg = dcg / idcg

    return ndcg


def calculate_mean(numbers):
    if not numbers:
        return 0.0
    return sum(numbers) / len(numbers)
```

```python
cosine_mxs = {
    "BKAI/Vietnamese-bi-encoder": vie_bi_encoder_cosine_sim,
    "AITeamVN/Vietnamese-Embedding": vie_embed_cosine_sim,
    "TF-IDF": cosine_sim,
}

for name, embedding in cosine_mxs.items():
    print("-------------------------------------")
    print(f"Evaluating model: {name}")
    rtopk_val = []
    rtopk_test = []
    ptopk_val = []
    ptopk_test = []
    mrr_val = []
    mrr_test = []
    ndcg_val = []
    ndcg_test = []

    for user in df_purchase["user_id"].unique():
        temp = df_purchase[df_purchase["user_id"] == user]
        x_val = temp.iloc[2, 1]  # index product_id for train
        y_val = temp.iloc[1, 1]  # index product_id for val
        y_val_pred = content_base_recommendation(x_val, embedding).tolist()
        rtopk_val.append(RtopK(y_val, y_val_pred))
        ptopk_val.append(PtopK(y_val, y_val_pred))
        mrr_val.append(MRR(y_val, y_val_pred))
        ndcg_val.append(NDCG(y_val, y_val_pred, 10))

        x_test = temp.iloc[1, 1]
        y_test = temp.iloc[0, 1]
        y_test_pred = content_base_recommendation(x_test, embedding).tolist()
        rtopk_test.append(RtopK(y_test, y_test_pred))
        ptopk_test.append(PtopK(y_test, y_test_pred))
        mrr_test.append(MRR(y_test, y_test_pred))
        ndcg_test.append(NDCG(y_test, y_test_pred, 10))

    print("rtopk val:", calculate_mean(rtopk_val))
    print("rtopk test:", calculate_mean(rtopk_test))
    print("ptopk val:", calculate_mean(ptopk_val))
    print("ptopk test:", calculate_mean(ptopk_test))
    print("mrr val:", calculate_mean(mrr_val))
    print("mrr test:", calculate_mean(mrr_test))
    print("ndcg val:", calculate_mean(ndcg_val))
    print("ndcg test:", calculate_mean(ndcg_test))
```

```python
def recommendation(product_id, src_score, cosine_similarity_matrix):

    idx = indices[product_id]  # Lấy index của sản phẩm dựa trên product_id
    sim_scores = list(enumerate(cosine_similarity_matrix[idx]))

    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    sim_scores = sim_scores[1:101]  # lay 100 san pham, bo qua san pham cua chinh no
    df_indices = [i[0] for i in sim_scores]  # tra ve danh sach index procuct_id

    popularity_scores = df_product.loc[df_indices, "popularity_score"]
    temp_df = pd.DataFrame(sim_scores, columns=["index", "score"])
    temp_df.index = temp_df["index"]
    temp_df.drop(columns=["index"], axis=1, inplace=True)

    score = temp_df["score"] * src_score + popularity_scores * (1 - src_score)
    score = score.sort_values(ascending=False)
    product_indices = score.index[:10]

    return product_indices
```

```python
from tqdm import tqdm

index = 0
df = pd.DataFrame(
    columns=[
        "score_weight",
        "popular_weight",
        "rtopk_val",
        "rtopk_test",
        "ptopk_val",
        "ptopk_test",
        "mrr_val",
        "mrr_test",
        "ndcg_val",
        "ndcg_test",
    ]
)

for scr in tqdm(score):
    rtop_val = []
    rtop_test = []
    ptop_val = []
    ptop_test = []
    mrr_val = []
    mrr_test = []
    ndcg_val = []
    ndcg_test = []

    for user in df_purchase["user_id"].unique():
        temp = df_purchase[df_purchase["user_id"] == user]
        x_val = temp.iloc[2, 1]
        y_val = temp.iloc[1, 1]
        y_val_pred = recommendation(x_val, scr, vie_embed_cosine_sim).tolist()
        rtopk_val.append(RtopK(y_val, y_val_pred))
        ptopk_val.append(PtopK(y_val, y_val_pred))
        mrr_val.append(MRR(y_val, y_val_pred))
        ndcg_val.append(NDCG(y_val, y_val_pred, 10))

        x_test = temp.iloc[1, 1]
        y_test = temp.iloc[0, 1]
        y_test_pred = recommendation(x_test, scr, vie_embed_cosine_sim).tolist()
        rtopk_test.append(RtopK(y_test, y_test_pred))
        ptopk_test.append(PtopK(y_test, y_test_pred))
        mrr_test.append(MRR(y_test, y_test_pred))
        ndcg_test.append(NDCG(y_test, y_test_pred, 10))

    dict_temp = {
        "score_weight": scr,
        "popular_weight": 1 - scr,
        "rtopk_val": calculate_mean(rtopk_val),
        "rtopk_test": calculate_mean(rtopk_test),
        "ptopk_val": calculate_mean(ptopk_val),
        "ptopk_test": calculate_mean(ptopk_test),
        "mrr_val": calculate_mean(mrr_val),
        "mrr_test": calculate_mean(mrr_test),
        "ndcg_val": calculate_mean(ndcg_val),
        "ndcg_test": calculate_mean(ndcg_test),
    }

    df = pd.concat([df, pd.DataFrame(dict_temp, index=[index])], ignore_index=True)
    index += 1
```
