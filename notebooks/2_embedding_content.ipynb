## TF-IDF

**TF-IDF** (viết tắt của Term Frequency-Inverse Document Frequency) là một kỹ thuật thống kê dạng số dùng để đánh giá tầm quan trọng của một từ trong một tài liệu hoặc một văn bản trong một tập hợp các tài liệu (corpus). Giá trị TF-IDF càng cao thì độ quan trọng của từ đó càng lớn trong tài liệu và trong cả tập hợp tài liệu.

```python
import torch
from sentence_transformers import SentenceTransformer

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")
```

### Information Retrieval

#### Leverage Text Embeddings for Information Retrieval

```python
from sklearn.preprocessing import normalize
from sklearn.metrics.pairwise import cosine_similarity


def search_products_embedding(query, top_n=10, model_name="bi-encoder"):
    """
    Tìm kiếm sản phẩm dựa trên truy vấn văn bản sử dụng embedding model

    Args:
        query (str): Câu truy vấn tìm kiếm
        top_n (int): Số lượng kết quả trả về
        model_name (str): Tên model embedding ("bi-encoder" hoặc "vie-embed")

    Returns:
        list: Danh sách sản phẩm phù hợp nhất
    """
    # Load model và embedding sản phẩm
    if model_name == "bi-encoder":
        vietnamese_model = vietnamese_bi_encoder
        max_seq_length = 256
        product_embeddings = (
            vie_bi_encoder_embed  # Đây là ma trận (n_sản_phẩm, embedding_dim)
        )
    else:  # "vie-embed"
        vietnamese_model = vietnamese_embedding
        max_seq_length = 2048
        product_embeddings = vie_embed

    # Encode câu truy vấn
    query_embedding = vietnamese_model.encode([query], convert_to_numpy=True)
    # query_embedding = normalize(query_embedding)

    # Normalize toàn bộ embedding sản phẩm (chỉ cần làm 1 lần nếu chưa)
    # product_embeddings = normalize(product_embeddings)

    # Tính cosine similarity giữa query và tất cả sản phẩm
    cosine_similarities = cosine_similarity(
        query_embedding, product_embeddings
    ).flatten()

    # Lấy top N kết quả giống nhất
    top_indices = cosine_similarities.argsort()[::-1][:top_n]

    # Trả về các sản phẩm tương ứng (df_product phải có sẵn, chỉ số trùng với embedding)
    return df_product.iloc[top_indices]
```

```python
query = "Sữa rửa mặt Pháp dưỡng ẩm."
result_search = search_products_embedding(
    query, top_n=10, model_name="vietnamese_embedding"
)
print(f"Kết quả tìm kiếm cho '{query}':")
display(result_search["origin"])
```
